# ACC coverage run (Mad Hatters, 6 Oct): Colab half

**What it does:** works the 232 rows that a single team holds and we have no road on (records 40–59). It looks for our **first road at the record length**, and each tie is worth **0.5 points** to us. It uses our full beam with the exact finish table and the exact shortener. Every path is checked on the official verifier, and the Stable line comes along with it.

**Before the first run:** put `coverage_run.py` in the *Mad Hatters ACC Shared Hub* folder (UPG-H sent it to you).

**Steps:** 1) connect Drive, 2) run. **Keep this tab open.** If Colab disconnects, choose Run all again; finished rows are skipped.

This half takes the even-numbered targets and your PC takes the odd ones, so the two never overlap. If the PC can't run it, set `part = 'all'`.

Results go to `COVERAGE/LINES_COVERAGE_colab.txt` in the hub, and UPG-H's check-in compares them with the live board for you to submit.

```python
#@title 1. Connect Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#@title 2. Run (about 11 hours; safe to stop and rerun, it resumes)
hours = 11 #@param {type:"number"}
part = 'colab' #@param ["colab", "all"]
tier = '1' #@param ["1", "2", "all"]
import pathlib, shutil, subprocess, sys
HUB = pathlib.Path('/content/drive/MyDrive/Mad Hatters ACC Shared Hub')
src = sorted(HUB.glob('coverage_run*.py')) or sorted(pathlib.Path('/content/drive/MyDrive').rglob('coverage_run*.py'))
assert src, 'Put coverage_run.py in the Mad Hatters ACC Shared Hub folder'
shutil.copy(src[-1], '/content/coverage_run.py')
cmd = [sys.executable, '/content/coverage_run.py', '--out', str(HUB / 'COVERAGE'), '--part', part, '--tier', tier, '--hours', str(hours)]
p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
for line in p.stdout:
    if not line.startswith('  '): print(line, end='', flush=True)
print('exit', p.wait())
```
